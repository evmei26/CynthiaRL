```python
import asyncio
import random
import logging
from poke_env.player import RandomPlayer
from poke_env import AccountConfiguration
from poke_env.player import Player
logging.basicConfig(level=logging.INFO)

suffix = random.randint(0,9999)
class CynthiaPlayer(Player):
    def choose_move(self, battle):
        if not battle.available_moves:
            return self.choose_random_move(battle)

        me, opp = battle.active_pokemon, battle.opponent_active_pokemon
        scored = [(self._score(m, me, opp), m) for m in battle.available_moves]
        best = max(s for s, _ in scored)
        return self.create_order(random.choice([m for s, m in scored if s == best]))

    def _score(self, move, me, opp):
        if move.category.name == "STATUS":
            # skip status moves that can't do anything useful
            if opp.status is not None:
                return -10
            return 30  

        mult = opp.damage_multiplier(move)
        #Penalize doing nothing
        if mult == 0:
            return -100  
        stab = 1.5 if move.type in me.types else 1.0
        return move.base_power * stab * mult * move.accuracy
    
async def main():
    
    rl_agent = RandomPlayer(
        account_configuration=AccountConfiguration(f"Clanker{suffix}", None),
        avatar="cyrus-gen4",
        max_concurrent_battles=1, 
        team=rl_team, 
        battle_format="gen4anythinggoes",
        log_level=logging.INFO,
        
    )
    cynthia_agent = CynthiaPlayer(
        account_configuration=AccountConfiguration(f"Cynthia{suffix}", None),
        avatar="101",
        max_concurrent_battles=1, 
        team=cynthia_dp_team, 
        battle_format="gen4anythinggoes",
        log_level=logging.INFO
    )

    print("Battle started! Simulating...")
    await rl_agent.battle_against(cynthia_agent, n_battles=20)

    print("\n--- Battle Finished Successfully! ---")
    print(f"Finished battles: {rl_agent.n_finished_battles}")
    print(f"Agent Wins: {rl_agent.n_won_battles}")
    print(f"Cynthia Wins: {cynthia_agent.n_won_battles}")

await main()
```
